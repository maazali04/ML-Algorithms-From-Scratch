# Terminologies

1. **Root Node**

    The very first node at the top of the tree containing 100% of your raw dataset before any splits are made.

2. **Decision/ Internal Node**

    Any middle point in the tree where a question is asked (e.g., $X_1 \le 2.5$). It splits the incoming data into two sub-branches.

3. **Leaf (Terminal) Node**

    The end of a branch where no more questions are asked. It holds the final predicted class (e.g., "Cat").

4.  **Splitting & Threshold ($t$)**

    The act of picking a single feature $X_j$ and a numerical value $t$ to cut the data into two groups:$$\text{Group 1: } X_j \le t \quad \text{vs.} \quad \text{Group 2: } X_j > t$$

5. **Node Impurity (Gini vs. Entropy)**

    Impurity is a measure of how mixed or messy a group of samples is:

    - Pure Node (Impurity = 0): A box containing only Cats.

    - Impure Node (High Impurity): A box containing an equal mix of 10 Cats, 10 Dogs, and 10 Rats.

    **Gini Impurity & Entropy** are just two different math formulas used to calculate this "messiness" score.

6. **Information Gain (Impurity Reduction)**

    The improvement in clarity you get after making a split.

    A decision tree tests every possible question it could ask and picks the one that gives the highest Information Gain (i.e., makes the resulting child boxes as pure as possible).

# Generalized Formulation: CART Split Optimization

1. **Feature Space & Candidate Set Definition**

    Let dataset $S = \{(\mathbf{x}_1, y_1), (\mathbf{x}_2, y_2), \dots, (\mathbf{x}_N, y_N)\}$, where $\mathbf{x}_i \in \mathbb{R}^d$ and $y_i \in \{0, 1, \dots, K-1\}$.
    
    For each feature $j \in \{1, 2, \dots, d\}$, sort the unique feature values:
    $$\text{sort}(\mathbf{x}_{:, j}) = [x_{(1), j}, x_{(2), j}, \dots, x_{(M_j), j}]$$
    Generate candidate split thresholds $T_j$:
    $$T_j = \left\{ t_m = \frac{x_{(m), j} + x_{(m+1), j}}{2} \;\middle\vert{}\; m = 1, 2, \dots, M_j - 1 \right\}$$
    
2. **Binary Partition Operator**
    
    For a candidate pair $(j, t)$ with $j \in \{1, \dots, d\}$ and $t \in T_j$, partition $S$ into orthogonal child sets:
    $$S_L(j, t) = \{(\mathbf{x}_i, y_i) \in S \mid x_{i, j} \le t\}$$
    $$S_R(j, t) = \{(\mathbf{x}_i, y_i) \in S \mid x_{i, j} > t\}$$
    
3. **Class Probability & Impurity Definitions**

    For any subset $A \subseteq S$:
    $$p_k(A) = \frac{1}{\vert{}A\vert{}} \sum_{(\mathbf{x}_i, y_i) \in A} \mathbb{I}(y_i = k), \quad \text{for } k \in \{0, 1, \dots, K-1\}$$
    $$\text{Gini}(A) = 1 - \sum_{k=0}^{K-1} p_k(A)^2$$
    $$\text{Entropy}(A) = -\sum_{k=0}^{K-1} p_k(A) \log_2 p_k(A) \quad \text{where } (0 \log_2 0 \equiv 0)$$
    
4. **Weighted Split Impurity & Information Gain**

    $$\text{Impurity}_{\text{split}}(S, j, t) = \frac{\vert{}S_L(j, t)\vert{}}{\vert{}S\vert{}} I(S_L(j, t)) + \frac{\vert{}S_R(j, t)\vert{}}{\vert{}S\vert{}} I(S_R(j, t))$$
    $$\Delta I(S, j, t) = I(S) - \text{Impurity}_{\text{split}}(S, j, t)$$
    (where $I(\cdot)$ is either $\text{Gini}(\cdot)$ or $\text{Entropy}(\cdot)$)
    
5. **Universal Argmax Optimization Rule**
    
    The optimal split feature index $j^*$ and threshold $t^*$ are determined by:
    $$(j^*, t^*) = \arg\max_{j \in \{1, \dots, d\}, \, t \in T_j} \Delta I(S, j, t)$$


**The Stopping Criteria (Base Cases)**

A tree cannot keep splitting forever, or it will assign every single sample to its own leaf (overfitting). A node becomes a Leaf Node (terminal prediction node) if ANY of these conditions are met:

$$\text{Create Leaf Node if: } \begin{cases}  1. & \text{Pure Node: } G(S) = 0 \text{ or } H(S) = 0 \\ 2. & \text{Max Depth Reached: } \text{depth} \ge d_{\max} \\ 3. & \text{Insufficient Samples: } \vert{}S\vert{} < N_{\min\_split} \\ 4. & \text{No Positive Gain: } \max_{j, t} \Delta I(S, j, t) \le 0 \end{cases}$$

When a node becomes a Leaf, its predicted value is the Majority Class:
$$\hat{y}_{\text{leaf}} = \arg\max_{k \in \{0, \dots, K-1\}} \sum_{i \in S} \mathbb{I}(y_i = k)$$

## Summary

1. **Start at the Root Node**
    - You start at the top with your entire dataset $S$. This is your **Root Node**.
    - You calculate its initial impurity score—either Gini or Entropy—to measure how mixed/messy the labels are.

2. **Sort Features & Find Candidate Thresholds**
    - For every single feature (e.g., Weight, Height, Ear Length):
        1. Sort all unique numerical values in ascending order.
        2. Compute the midpoint between every adjacent pair of values.
        3. These midpoints become your list of candidate thresholds ($t$).

3. **Test Every Single Split (The Double Loop)**
    - For every candidate threshold on every feature, split the data into a Left Child ($X_j \le t$) and a Right Child ($X_j > t$).
    - Calculate the Weighted Child Impurity ($G_{\text{split}}$):
    $$G_{\text{split}} = \left(\frac{N_L}{N} \times G_{\text{Left}}\right) + \left(\frac{N_R}{N} \times G_{\text{Right}}\right)$$
    - Calculate the Information Gain ($\Delta G$):$$\Delta G = \text{Initial Impurity} - G_{\text{split}}$$

4. **Pick the Winner (Argmax Selection)**
    - Compare the Information Gain of every threshold across every feature.
    - Select the single (feature, threshold) pair that produced the highest Information Gain ($\arg\max \Delta G$).
    - If there's a tie, pick the first one encountered!

5. **Repeat Recursively Until a Stopping Rule Hits**
    - Send the left and right split datasets down to child nodes and repeat Steps 1–4 for each branch.
    - Stop splitting a branch and turn it into a Leaf Node if:
        1. The node is 100% pure ($G=0$).
        2. You reach the Maximum Depth limit.
        3. The number of samples falls below the Minimum Split threshold.
        4. No split provides a positive Gain ($\Delta G \le 0$).

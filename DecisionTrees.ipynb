# Terminologies

1. **Root Node**

    The very first node at the top of the tree containing 100% of your raw dataset before any splits are made.

2. **Decision/ Internal Node**

    Any middle point in the tree where a question is asked (e.g., $X_1 \le 2.5$). It splits the incoming data into two sub-branches.

3. **Leaf (Terminal) Node**

    The end of a branch where no more questions are asked. It holds the final predicted class (e.g., "Cat").

4.  **Splitting & Threshold ($t$)**

    The act of picking a single feature $X_j$ and a numerical value $t$ to cut the data into two groups:$$\text{Group 1: } X_j \le t \quad \text{vs.} \quad \text{Group 2: } X_j > t$$

5. **Node Impurity (Gini vs. Entropy)**

    Impurity is a measure of how mixed or messy a group of samples is:

    - Pure Node (Impurity = 0): A box containing only Cats.

    - Impure Node (High Impurity): A box containing an equal mix of 10 Cats, 10 Dogs, and 10 Rats.

    **Gini Impurity & Entropy** are just two different math formulas used to calculate this "messiness" score.

6. **Information Gain (Impurity Reduction)**

    The improvement in clarity you get after making a split.

    A decision tree tests every possible question it could ask and picks the one that gives the highest Information Gain (i.e., makes the resulting child boxes as pure as possible).

## Generalized Formulation: CART Split Optimization

1. **Feature Space & Candidate Set Definition**

    Let dataset $S = \{(\mathbf{x}_1, y_1), (\mathbf{x}_2, y_2), \dots, (\mathbf{x}_N, y_N)\}$, where $\mathbf{x}_i \in \mathbb{R}^d$ and $y_i \in \{0, 1, \dots, K-1\}$.
    
    For each feature $j \in \{1, 2, \dots, d\}$, sort the unique feature values:
    $$\text{sort}(\mathbf{x}_{:, j}) = [x_{(1), j}, x_{(2), j}, \dots, x_{(M_j), j}]$$
    Generate candidate split thresholds $T_j$:
    $$T_j = \left\{ t_m = \frac{x_{(m), j} + x_{(m+1), j}}{2} \;\middle\vert{}\; m = 1, 2, \dots, M_j - 1 \right\}$$
    
2. **Binary Partition Operator**
    
    For a candidate pair $(j, t)$ with $j \in \{1, \dots, d\}$ and $t \in T_j$, partition $S$ into orthogonal child sets:
    $$S_L(j, t) = \{(\mathbf{x}_i, y_i) \in S \mid x_{i, j} \le t\}$$
    $$S_R(j, t) = \{(\mathbf{x}_i, y_i) \in S \mid x_{i, j} > t\}$$
    
3. **Class Probability & Impurity Definitions**

    For any subset $A \subseteq S$:
    $$p_k(A) = \frac{1}{\vert{}A\vert{}} \sum_{(\mathbf{x}_i, y_i) \in A} \mathbb{I}(y_i = k), \quad \text{for } k \in \{0, 1, \dots, K-1\}$$
    $$\text{Gini}(A) = 1 - \sum_{k=0}^{K-1} p_k(A)^2$$
    $$\text{Entropy}(A) = -\sum_{k=0}^{K-1} p_k(A) \log_2 p_k(A) \quad \text{where } (0 \log_2 0 \equiv 0)$$
    
4. **Weighted Split Impurity & Information Gain**

    $$\text{Impurity}_{\text{split}}(S, j, t) = \frac{\vert{}S_L(j, t)\vert{}}{\vert{}S\vert{}} I(S_L(j, t)) + \frac{\vert{}S_R(j, t)\vert{}}{\vert{}S\vert{}} I(S_R(j, t))$$
    $$\Delta I(S, j, t) = I(S) - \text{Impurity}_{\text{split}}(S, j, t)$$
    (where $I(\cdot)$ is either $\text{Gini}(\cdot)$ or $\text{Entropy}(\cdot)$)
    
5. **Universal Argmax Optimization Rule**
    
    The optimal split feature index $j^*$ and threshold $t^*$ are determined by:
    $$(j^*, t^*) = \arg\max_{j \in \{1, \dots, d\}, \, t \in T_j} \Delta I(S, j, t)$$


**The Stopping Criteria (Base Cases)**

A tree cannot keep splitting forever, or it will assign every single sample to its own leaf (overfitting). A node becomes a Leaf Node (terminal prediction node) if ANY of these conditions are met:

$$\text{Create Leaf Node if: } \begin{cases}  1. & \text{Pure Node: } G(S) = 0 \text{ or } H(S) = 0 \\ 2. & \text{Max Depth Reached: } \text{depth} \ge d_{\max} \\ 3. & \text{Insufficient Samples: } \vert{}S\vert{} < N_{\min\_split} \\ 4. & \text{No Positive Gain: } \max_{j, t} \Delta I(S, j, t) \le 0 \end{cases}$$

When a node becomes a Leaf, its predicted value is the Majority Class:
$$\hat{y}_{\text{leaf}} = \arg\max_{k \in \{0, \dots, K-1\}} \sum_{i \in S} \mathbb{I}(y_i = k)$$

### Summary

1. **Start at the Root Node**
    - You start at the top with your entire dataset $S$. This is your **Root Node**.
    - You calculate its initial impurity score—either Gini or Entropy—to measure how mixed/messy the labels are.

2. **Sort Features & Find Candidate Thresholds**
    - For every single feature (e.g., Weight, Height, Ear Length):
        1. Sort all unique numerical values in ascending order.
        2. Compute the midpoint between every adjacent pair of values.
        3. These midpoints become your list of candidate thresholds ($t$).

3. **Test Every Single Split (The Double Loop)**
    - For every candidate threshold on every feature, split the data into a Left Child ($X_j \le t$) and a Right Child ($X_j > t$).
    - Calculate the Weighted Child Impurity ($G_{\text{split}}$):
    $$G_{\text{split}} = \left(\frac{N_L}{N} \times G_{\text{Left}}\right) + \left(\frac{N_R}{N} \times G_{\text{Right}}\right)$$
    - Calculate the Information Gain ($\Delta G$):$$\Delta G = \text{Initial Impurity} - G_{\text{split}}$$

4. **Pick the Winner (Argmax Selection)**
    - Compare the Information Gain of every threshold across every feature.
    - Select the single (feature, threshold) pair that produced the highest Information Gain ($\arg\max \Delta G$).
    - If there's a tie, pick the first one encountered!

5. **Repeat Recursively Until a Stopping Rule Hits**
    - Send the left and right split datasets down to child nodes and repeat Steps 1–4 for each branch.
    - Stop splitting a branch and turn it into a Leaf Node if:
        1. The node is 100% pure ($G=0$).
        2. You reach the Maximum Depth limit.
        3. The number of samples falls below the Minimum Split threshold.
        4. No split provides a positive Gain ($\Delta G \le 0$).

## Python Code

```python
import numpy as np

class DecisionTreeClassifier:

    def __init__(self, max_depth=3, min_samples_split=2):
        self.max_depth = max_depth
        self.min_samples_split = min_samples_split
        self.tree_blueprint = None

    def _calculate_gini(self, y_val):
        if len(y_val) == 0:
            return 0.0
        values, counts = np.unique(y_val, return_counts=True)
        proba = counts / len(y_val)
        gini = 1.0 - np.sum(proba ** 2)
        return gini

    def _find_best_split(self, X_node, y_node):
        best_gain = -1.0
        best_thresh = None
        thresh_col = None
        total_sample = len(y_node)
        current_node_gini = self._calculate_gini(y_node)
        features_count = X_node.shape[1]

        for feature in range(features_count):
            sorted_indices = np.argsort(X_node[:, feature])
            feature_sort = X_node[:, feature][sorted_indices]
            label_sort = y_node[sorted_indices]

            thresholds = [(feature_sort[idx] + feature_sort[idx + 1]) / 2.0 
                          for idx in range(len(feature_sort) - 1)]

            for t in thresholds:
                right_label = label_sort[feature_sort > t]
                left_label = label_sort[feature_sort <= t]

                if len(right_label) == 0 or len(left_label) == 0:
                    continue

                right_gini = self._calculate_gini(right_label)
                left_gini = self._calculate_gini(left_label)

                G_split = (len(right_label) / total_sample) * right_gini + \
                          (len(left_label) / total_sample) * left_gini
                
                gain = current_node_gini - G_split

                if gain > best_gain:
                    best_gain = gain
                    best_thresh = t
                    thresh_col = feature

        return thresh_col, best_thresh, best_gain

    def _recursion(self, X_node, y_node, depth=0):
        node_samples = len(y_node)
        current_gini = self._calculate_gini(y_node)

        # 1. Pure Node
        if current_gini == 0.0:
            majority_class = np.bincount(y_node).argmax()
            return {"type": "leaf", "value": majority_class, "reason": "Pure Node"}

        # 2. Max Depth
        if depth >= self.max_depth:
            majority_class = np.bincount(y_node).argmax()
            return {"type": "leaf", "value": majority_class, "reason": "Max Depth"}

        # 3. Insufficient Samples
        if node_samples < self.min_samples_split:
            majority_class = np.bincount(y_node).argmax()
            return {"type": "leaf", "value": majority_class, "reason": "Insufficient Samples"}

        # Find best feature & threshold split
        col, thresh, gain = self._find_best_split(X_node, y_node)

        # 4. No Positive Gain
        if gain <= 0.0 or thresh is None:
            majority_class = np.bincount(y_node).argmax()
            return {"type": "leaf", "value": majority_class, "reason": "No Positive Gain"}

        left_mask = X_node[:, col] <= thresh
        right_mask = X_node[:, col] > thresh

        
        left_branch = self._recursion(X_node[left_mask], y_node[left_mask], depth + 1)
        right_branch = self._recursion(X_node[right_mask], y_node[right_mask], depth + 1)

        return {
            "type": "split",
            "feature_index": col,
            "threshold": thresh,
            "left": left_branch,
            "right": right_branch
        }

    def _predict_sample(self, sample, tree_node):
        if tree_node["type"] == "leaf":
            return tree_node["value"]

        feature_idx = tree_node["feature_index"]
        threshold_val = tree_node["threshold"]

        if sample[feature_idx] <= threshold_val:
            return self._predict_sample(sample, tree_node["left"])
        else:
            return self._predict_sample(sample, tree_node["right"])

    def fit(self, X, y):
        self.tree_blueprint = self._recursion(X, y, depth=0)

    def predict(self, X):
        predictions = [self._predict_sample(sample, self.tree_blueprint) for sample in X]
        return np.array(predictions)
```

```python
# Animal Dataset Matrix
X = np.array([
    [12.0, 4.0],  # Dog
    [15.0, 5.0],  # Dog
    [3.0, 3.5],   # Cat
    [4.0, 4.5],   # Cat
    [0.5, 1.0],   # Rat
    [0.8, 1.2]    # Rat
])
y = np.array([1, 1, 0, 0, 2, 2])

# Train Model
clf = DecisionTreeClassifier(max_depth=3)
clf.fit(X, y)

# Predict
preds = clf.predict(X)
print("Predictions:", preds)
# Output: Predictions: [1 1 0 0 2 2]
```

## Q&A

**Question**: Why is it Called CART?

**Answer**: CART stands for Classification And Regression Trees.
The name perfectly describes its dual nature (Classification Trees, Regression Trees)

---
**Question**: Can a tree split into 3, 4, or 10 branches at a single node?

**Answer**: Yes, absolutely!
What we coded and derived is called a **Binary Split** (used by the **CART algorithm**). But other algorithms perform Multiway (N-ary) Splits (Instead of asking a binary question ($X \le 8.0$), the algorithm evaluates categorical features with multiple discrete states (e.g., $\text{Color} \in \{\text{Red}, \text{Green}, \text{Blue}\}$) and splits the parent node directly into 3 separate child branches in a single step.) 

---
**Question**: Can a single node check MULTIPLE rules at once?

**Answer**: Yes, this is called an Oblique (or Multivariate) Decision Tree!
Instead of checking a single axis-aligned feature, an oblique node checks a linear combination of multiple features at the same time using a formula like: 
$$w_{1}\cdot \text{Weight}+w_{2}\cdot \text{Ear\ Length}\le \text{Threshold}$$

---

**Question**: What other types of Decision Tree algorithms exist besides CART?

**Answer**: CART (Classification And Regression Trees) is just one family. Here are the main historic and modern tree families:

|Algorithm|Split Type|Impurity Metric|Target Support|Key Characteristics|
|---|---|---|---|---|
|CART |Strictly Binary|Gini Impurity (Class) or MSE (Reg)|Classification & Regression|Used by scikit-learn, xgboost, LightGBM.|
|ID3 (Iterative Dichotomiser 3)|Multiway|Entropy / Information Gain|Classification Only|Created by Ross Quinlan. Only works on categorical features.|
|C4.5 (Successor to ID3)|Multiway & Binary|Gain Ratio (normalizes Information Gain)|Classification Only|Handles continuous attributes, missing data values, and post-pruning.|
|C5.0|Multiway|Gain Ratio|Classification Only|Commercial variant of C4.5; faster and memory-efficient.|
|CHAID (Chi-square Automatic Interaction Detection)|Multiway|Chi-Square $\chi^2$ Test (p-values)|Classification & Regression|Used heavily in statistics and marketing research; stops splitting using statistical significance.|

# CART Regresssion

- Target Variable: Continuous numerical values.
- Node Impurity Metric: MSE or variance
- Leaf Prediction: Sample Mean
- Split Optimization: Maximize Variance Reduction


**Mechanism**

1. Compute Mean ȳ (Baseline Prediction)

$$\bar{y}_S = \frac{1}{\vert{}S\vert{}} \sum_{i \in S} y_i$$

2. Compute Node MSE (Total Variance / Impurity)

$$\text{MSE}(S) = \frac{1}{\vert{}S\vert{}} \sum_{i \in S} (y_i - \bar{y}_S)^2$$

3. Test All Split Candidates (Features & Thresholds)

4. Compute Split MSE & Variance Reduction (ΔMSE)

    $$S_L = \{(\mathbf{x}_i, y_i) \in S \mid x_{i,j} \le t\}$$
    $$S_R = \{(\mathbf{x}_i, y_i) \in S \mid x_{i,j} > t\}$$
    Weighted Child Impurity ($\text{MSE}_{\text{split}}$)
    $$\text{MSE}_{\text{split}}(S, j, t) = \left( \frac{\vert{}S_L\vert{}}{\vert{}S\vert{}} \times \text{MSE}(S_L) \right) + \left( \frac{\vert{}S_R\vert{}}{\vert{}S\vert{}} \times \text{MSE}(S_R) \right)$$
    $$\Delta \text{MSE}(S, j, t) = \text{MSE}(S) - \text{MSE}_{\text{split}}(S, j, t)$$

5. Pick (j*, t*) with Maximum Variance Reduction & Split
$$(j^*, t^*) = \arg\max_{j \in \{1, \dots, d\}, \, t \in T_j} \Delta \text{MSE}(S, j, t)$$

**Note**: When a stopping rule is triggered (e.g., node variance reaches 0, max depth is reached, or $\vert{}S\vert{} < N_{\min\_split}$), the node becomes a Leaf Node.Instead of taking a majority vote, the continuous prediction for any new input $\mathbf{x}_{\text{new}}$ that lands in that leaf is the sample mean of the training targets contained in that leaf:$$\hat{y}_{\text{leaf}} = \bar{y}_{S_{\text{leaf}}} = \frac{1}{\vert{}S_{\text{leaf}}\vert{}} \sum_{i \in S_{\text{leaf}}} y_i$$


```python
import numpy as np

class DecisionTreeRegressor:

    def __init__(self, max_depth=3, min_samples_split=2):
        self.max_depth = max_depth
        self.min_samples_split = min_samples_split
        self.tree_blueprint = None

    def _calculate_mse(self, y_val):
        """
        REGRESSION SHIFT: Measures variance/MSE instead of Gini.
        MSE is the average squared difference from the mean.
        """
        if len(y_val) == 0:
            return 0.0
        
        # Calculate the mean (average) of the target numbers in this node
        node_mean = np.mean(y_val)
        
        # Compute Mean Squared Error (Variance)
        mse = np.mean((y_val - node_mean) ** 2)
        return mse

    def _find_best_split(self, X_node, y_node):
        best_variance_reduction = -1.0
        best_thresh = None
        thresh_col = None
        total_sample = len(y_node)
        
        # Current baseline MSE before splitting
        current_node_mse = self._calculate_mse(y_node)
        features_count = X_node.shape[1]

        for feature in range(features_count):
            sorted_indices = np.argsort(X_node[:, feature])
            feature_sort = X_node[:, feature][sorted_indices]
            label_sort = y_node[sorted_indices]

            thresholds = [(feature_sort[idx] + feature_sort[idx + 1]) / 2.0 
                          for idx in range(len(feature_sort) - 1)]

            for t in thresholds:
                right_label = label_sort[feature_sort > t]
                left_label = label_sort[feature_sort <= t]

                if len(right_label) == 0 or len(left_label) == 0:
                    continue

                right_mse = self._calculate_mse(right_label)
                left_mse = self._calculate_mse(left_label)

                # Weight the children's MSE based on sample count
                weighted_mse = (len(right_label) / total_sample) * right_mse + \
                               (len(left_label) / total_sample) * left_mse
                
                # REGRESSION SHIFT: Information Gain is now "Variance Reduction"
                variance_reduction = current_node_mse - weighted_mse

                if variance_reduction > best_variance_reduction:
                    best_variance_reduction = variance_reduction
                    best_thresh = t
                    thresh_col = feature

        return thresh_col, best_thresh, best_variance_reduction

    def _recursion(self, X_node, y_node, depth=0):
        node_samples = len(y_node)
        current_mse = self._calculate_mse(y_node)

        # REGRESSION SHIFT: Leaf value is the statistical MEAN of the targets
        node_prediction = np.mean(y_node) if len(y_node) > 0 else 0.0

        # Stop Rule 1: Perfectly pure regression node (all continuous values are identical)
        if current_mse == 0.0:
            return {"type": "leaf", "value": node_prediction, "reason": "Pure Node (Zero Variance)"}

        # Stop Rule 2: Max Depth
        if depth >= self.max_depth:
            return {"type": "leaf", "value": node_prediction, "reason": "Max Depth"}

        # Stop Rule 3: Insufficient Samples
        if node_samples < self.min_samples_split:
            return {"type": "leaf", "value": node_prediction, "reason": "Insufficient Samples"}

        col, thresh, variance_red = self._find_best_split(X_node, y_node)

        # Stop Rule 4: No Reduction in Variance
        if variance_red <= 0.0 or thresh is None:
            return {"type": "leaf", "value": node_prediction, "reason": "No Variance Reduction"}

        left_mask = X_node[:, col] <= thresh
        right_mask = X_node[:, col] > thresh
        
        left_branch = self._recursion(X_node[left_mask], y_node[left_mask], depth + 1)
        right_branch = self._recursion(X_node[right_mask], y_node[right_mask], depth + 1)

        return {
            "type": "split",
            "feature_index": col,
            "threshold": thresh,
            "left": left_branch,
            "right": right_branch
        }

    def _predict_sample(self, sample, tree_node):
        if tree_node["type"] == "leaf":
            return tree_node["value"]

        feature_idx = tree_node["feature_index"]
        threshold_val = tree_node["threshold"]

        if sample[feature_idx] <= threshold_val:
            return self._predict_sample(sample, tree_node["left"])
        else:
            return self._predict_sample(sample, tree_node["right"])

    def fit(self, X, y):
        self.tree_blueprint = self._recursion(X, y, depth=0)

    def predict(self, X):
        predictions = [self._predict_sample(sample, self.tree_blueprint) for sample in X]
        return np.array(predictions)
```

```python
# 1. Continuous Dataset: [Square Footage (k sqft), House Age (years)] -> Price ($k)
X_train = np.array([
    [1.0, 10], # $200k
    [1.2, 8],  # $220k
    [2.0, 5],  # $350k
    [2.5, 2],  # $410k
    [3.0, 1]   # $500k
])
y_train = np.array([200.0, 220.0, 350.0, 410.0, 500.0])

# 2. Fit Regressor
reg = DecisionTreeRegressor(max_depth=3, min_samples_split=2)
reg.fit(X_train, y_train)

# 3. Predict on Training Data
preds = reg.predict(X_train)

# 4. Compute Regression Metrics (MSE & R²)
mse = np.mean((y_train - preds) ** 2)
ss_tot = np.sum((y_train - np.mean(y_train)) ** 2)
ss_res = np.sum((y_train - preds) ** 2)
r2_score = 1 - (ss_res / ss_tot)

print("Predictions ($k):", preds)
print(f"Mean Squared Error: {mse:.4f}")
print(f"R² Score: {r2_score:.4f}")
```

# Cost-Complexity Pruning (CCP)

Decision Trees are notoriously prone to overfitting. Left unrestricted, a tree will continue splitting until every leaf node contains a single sample ($Gini = 0$ or $MSE = 0$). While this achieves $100\%$ accuracy on training data, it fails to generalize to unseen test data.


There are two main strategies to combat this:

1. **Pre-Pruning (Early Stopping)**: Stopping the tree during growth using hyperparameters like max_depth or min_samples_split.

2. **Post-Pruning (Cost-Complexity Pruning)**: Growing a full, unconstrained tree first, then systematically removing (pruning) non-essential subtrees from the bottom up.

## CCP Mechanism

Cost-Complexity Pruning (CCP) only runs at the very end, after the tree is fully grown, but it evaluates the nodes one by one to figure out where to chop.


1.  **Grow the Tree Fully**
    You let the tree grow completely until it cannot split anymore.

2. **Look at an Internal Node**
    Now the tree stops growing, and the CCP algorithm starts working. It doesn't look at the leaves; it looks at the internal nodes (the junctions where a split happens).

    Let's suppose we have a **Node t**. Beneath Node t, we have 5 splits and 10 leaf nodes. Now CCP aska a 'What-If' question about this specific node:

    "What happens to our error if I cut off everything below Node t, turning Node t into a flat, final leaf node right now?"

3. **Compare the "Keep" vs. "Cut" Errors**

    To answer that question, the algorithm calculates two error scors just for that section of the data:
    1. **$R(T_t)$(The "Keep" Error)**: This is the error rate if you keep the whole messy branch beneath Node t. 
    2. **$R(t)$(The "Cut" Error)**: This is the error rate if yo uchop off the branch and turn Node t into a leaf. Because Node t contains a mix of different animals, making it a leaf forces it to make a single majority guess. Therefore, this error rises above 0.

4. **Calculate the "Cost-per-Leaf" (g(t))**

    Now, the algorithm calculates the Effective Alpha g(t) for this node. This score tells the computer: "How much does our training error increase for every single leaf node we destroy by cutting here?"

    The formula is:
    $$g(t)=\frac{\text{Error\ when\ Cut\ }R(t)-\text{Error\ when\ Kept\ }R(T_{t})}{\text{Number\ of\ Leaves\ Saved}-1}$$

    The computer calculates this g(t) score for every single internal node in the entire tree.


5. **Prune the Weakest Node First**

    The computer looks at its list of g(t) scores for all nodes.
    
    A low g(t) score means: "If you cut here, your error barely goes up, but you save a lot of complex leaves." This is a weak, overfitted branch.
    
    The computer finds the node with the absolute lowest g(t) score and chops it off. That node permanently becomes a leaf.
    
6. **Loop Until Nothing Is Left**

    Now that the tree is slightly smaller, the computer updates its math and repeats the whole process:
    1. It recalculates the g(t) scores for the remaining nodes.
    2. It finds the next lowest score.
    3. It chops that branch off.

    It keeps looping until the entire tree is chopped all the way down to a single root node.

7. **Select the optimal $\alpha$ using a validation set or Cross-Validation.**
    CCP does not know when to stop on its own. It stops where you tell it to stop, based on which pruned variation achieves the best balance and highest accuracy on unseen test data.
    $$R_\alpha(T) = R(T) + \alpha \cdot \vert{}T\vert{}$$
    Where:
    - $R(T)$ is the Total Training Impurity of the tree (sum of weighted Gini/MSE across all leaf nodes):
    $$R(T) = \sum_{l \in \text{Leaves}(T)} \vert{}S_l\vert{} \cdot I(S_l)$$
    - $\vert{}T\vert{}$ is the Number of Leaf Nodes in tree $T$.
    - $\alpha \ge 0$ is the Complexity Parameter (the regularization multiplier):
        - $\alpha = 0$: No penalty. The tree remains fully grown (complex, overfitted).
        - $\alpha \to \infty$: Infinite penalty. The tree is pruned all the way down to a single root node.

```python
import numpy as np
import copy

class DecisionTree:
    """Full CART Decision Tree with helper functions for CCP."""

    def __init__(self, max_depth=None, min_samples_split=2):
        self.max_depth = max_depth if max_depth is not None else float('inf')
        self.min_samples_split = min_samples_split
        self.tree_blueprint = None

    def _calculate_gini(self, y_val):
        if len(y_val) == 0:
            return 0.0
        _, counts = np.unique(y_val, return_counts=True)
        proba = counts / len(y_val)
        return 1.0 - np.sum(proba ** 2)

    def _find_best_split(self, X_node, y_node):
        best_gain = -1.0
        best_thresh = None
        thresh_col = None
        total_sample = len(y_node)
        current_node_gini = self._calculate_gini(y_node)

        for feature in range(X_node.shape[1]):
            sorted_indices = np.argsort(X_node[:, feature])
            feature_sort = X_node[:, feature][sorted_indices]
            label_sort = y_node[sorted_indices]

            thresholds = [(feature_sort[i] + feature_sort[i + 1]) / 2.0 
                          for i in range(len(feature_sort) - 1)]

            for t in thresholds:
                right_label = label_sort[feature_sort > t]
                left_label = label_sort[feature_sort <= t]

                if len(right_label) == 0 or len(left_label) == 0:
                    continue

                G_split = (len(right_label) / total_sample) * self._calculate_gini(right_label) + \
                          (len(left_label) / total_sample) * self._calculate_gini(left_label)
                
                gain = current_node_gini - G_split

                if gain > best_gain:
                    best_gain = gain
                    best_thresh = t
                    thresh_col = feature

        return thresh_col, best_thresh, best_gain

    # STEP 1: Grow the Tree Fully
    def _grow_tree(self, X_node, y_node, depth=0, node_id=0):
        node_samples = len(y_node)
        current_gini = self._calculate_gini(y_node)
        majority_class = np.bincount(y_node).argmax() if len(y_node) > 0 else 0

        # Create base dictionary node
        node = {
            "node_id": node_id,
            "type": "leaf",
            "value": majority_class,
            "gini": current_gini,
            "n_samples": node_samples
        }

        # Stopping Criteria
        if current_gini == 0.0 or depth >= self.max_depth or node_samples < self.min_samples_split:
            return node

        col, thresh, gain = self._find_best_split(X_node, y_node)

        if gain <= 0.0 or thresh is None:
            return node

        left_mask = X_node[:, col] <= thresh
        right_mask = X_node[:, col] > thresh

        # Convert node to split node
        node["type"] = "split"
        node["feature_index"] = col
        node["threshold"] = thresh
        node["left"] = self._grow_tree(X_node[left_mask], y_node[left_mask], depth + 1, node_id * 2 + 1)
        node["right"] = self._grow_tree(X_node[right_mask], y_node[right_mask], depth + 1, node_id * 2 + 2)

        return node

    def fit(self, X, y):
        self.tree_blueprint = self._grow_tree(X, y, depth=0, node_id=0)

    def _predict_sample(self, sample, tree_node):
        if tree_node["type"] == "leaf":
            return tree_node["value"]

        if sample[tree_node["feature_index"]] <= tree_node["threshold"]:
            return self._predict_sample(sample, tree_node["left"])
        else:
            return self._predict_sample(sample, tree_node["right"])

    def predict(self, X):
        return np.array([self._predict_sample(sample, self.tree_blueprint) for sample in X])


class CostComplexityPruning:
    """Implements CCP using the 7-step sequence you described."""

    def __init__(self, model):
        self.model = model

    # STEPS 2, 3, & 4: Calculate R(T_t), R(t), and Effective Alpha g(t) for every node
    def _get_subtree_metrics(self, node):
        """Returns total weighted impurity R(T_t) and leaf count |T_t| for a subtree."""
        if node["type"] == "leaf":
            return node["n_samples"] * node["gini"], 1

        left_R, left_leaves = self._get_subtree_metrics(node["left"])
        right_R, right_leaves = self._get_subtree_metrics(node["right"])

        return left_R + right_R, left_leaves + right_leaves

    def _compute_gt_for_all_nodes(self, node, gt_dict):
        """Calculates g(t) = [R(t) - R(T_t)] / (|T_t| - 1) for all internal nodes."""
        if node["type"] == "leaf":
            return

        # 1. R(T_t): "Keep" Error (Subtree Error)
        r_subtree, n_leaves = self._get_subtree_metrics(node)

        # 2. R(t): "Cut" Error (Node Impurity if collapsed into a leaf)
        r_node = node["n_samples"] * node["gini"]

        # 3. g(t): Cost-per-leaf saved
        gt = (r_node - r_subtree) / (n_leaves - 1) if n_leaves > 1 else 0.0
        gt_dict[node["node_id"]] = gt

        # Recursively check children internal nodes
        self._compute_gt_for_all_nodes(node["left"], gt_dict)
        self._compute_gt_for_all_nodes(node["right"], gt_dict)

    # STEP 5: Prune the Weakest Node First (Lowest g(t))
    def _prune_node_by_id(self, node, target_id):
        """Traverses the tree and turns the node with target_id into a leaf."""
        if node["type"] == "leaf":
            return node

        if node["node_id"] == target_id:
            # Turn this internal split node into a Leaf Node!
            return {
                "node_id": node["node_id"],
                "type": "leaf",
                "value": node["value"],
                "gini": node["gini"],
                "n_samples": node["n_samples"]
            }

        node["left"] = self._prune_node_by_id(node["left"], target_id)
        node["right"] = self._prune_node_by_id(node["right"], target_id)
        return node

    # STEPS 6 & 7: Loop until only Root Node remains, then pick optimal tree using Validation Data
    def prune_and_select_best(self, X_val, y_val):
        current_tree = copy.deepcopy(self.model.tree_blueprint)
        
        candidate_trees = []
        
        # Step 6: Iteratively find lowest g(t) and chop it off
        while current_tree["type"] != "leaf":
            gt_dict = {}
            self._compute_gt_for_all_nodes(current_tree, gt_dict)

            if not gt_dict:
                break

            # Find target node with minimum g(t) score
            weakest_node_id = min(gt_dict, key=gt_dict.get)
            
            # Save candidate tree blueprint
            candidate_trees.append(copy.deepcopy(current_tree))

            # Prune weakest node
            current_tree = self._prune_node_by_id(current_tree, weakest_node_id)

        # Add final root leaf tree to candidates
        candidate_trees.append(copy.deepcopy(current_tree))

        # Step 7: Evaluate each candidate tree on Validation Data
        best_accuracy = -1.0
        best_tree_blueprint = None

        for idx, tree_candidate in enumerate(candidate_trees):
            self.model.tree_blueprint = tree_candidate
            preds = self.model.predict(X_val)
            acc = np.mean(preds == y_val)

            if acc > best_accuracy:
                best_accuracy = acc
                best_tree_blueprint = tree_candidate

        # Assign the winning tree back to the model
        self.model.tree_blueprint = best_tree_blueprint
        print(f"CCP complete! Generated {len(candidate_trees)} candidate trees.")
        print(f"Selected tree achieving optimal Validation Accuracy: {best_accuracy * 100:.2f}%")
        return self.model
```

```python
# 1. Create artificial dataset with noise
np.random.seed(42)
X_all = np.random.rand(150, 2) * 10
y_all = (X_all[:, 0] + X_all[:, 1] > 10).astype(int)
y_all[::8] = 1 - y_all[::8] # Introduce 12% label noise

# Split into Train (100 samples) and Validation (50 samples)
X_train, y_train = X_all[:100], y_all[:100]
X_val, y_val = X_all[100:], y_all[100:]

# 2. Train Unpruned Full Tree
tree = DecisionTree(max_depth=None, min_samples_split=2)
tree.fit(X_train, y_train)

print(f"Unpruned Tree Training Accuracy:   {np.mean(tree.predict(X_train) == y_train) * 100:.2f}%")
print(f"Unpruned Tree Validation Accuracy: {np.mean(tree.predict(X_val) == y_val) * 100:.2f}%")

print("\n--- Running Cost-Complexity Pruning (CCP) ---")
ccp = CostComplexityPruning(tree)
best_tree_model = ccp.prune_and_select_best(X_val, y_val)

print(f"Pruned Tree Validation Accuracy:   {np.mean(best_tree_model.predict(X_val) == y_val) * 100:.2f}%")
```

**Support Vector Machines (SVM)** are geometric, discriminative classifiers that separate data classes by finding an optimal decision boundary called a hyperplane. Unlike algorithms that find any line separating classes, SVM calculates the unique hyperplane that maximizes the distance—known as the margin—to the nearest data points.

SVM is used for classification as well as regression.


# Core Intuition


Imagine plotting your dataset on a 2D graph, where Red dots represent Spam emails and Blue dots represent Ham emails.

1. **Hyperplane**

    A hyperplane is the mathematical decision boundary that splits the data space into two halves. 

2. **Problem (Infinite Separating Lines)**
    
    If the Red and Blue clusters are well-separated, you can draw hundreds of different valid lines (hyperplane) that split them with $100\%$ accuracy.

    However, if you draw a line that sits too close to the Red dots, a new test sample that is just slightly shifted might end up on the wrong side of the line, causing a misclassification.

3. **Solution (Defining the Margin)**

    To pick the absolute best line out of the infinite possibilities, SVM measures the margin:
    
    - **Margin**: It is calculated as the Euclidean distance between the decision line and the closest data point on either side.

    - **Support Vectors:** Support vectors are those points that are closest to the hyperplane and the marginal line hits them. They are called "support vectors" because they physically hold up the position of the boundary—if you delete any other point in the dataset, the line stays identical.

4. **The Core Goal of SVM**

    SVM chooses the unique hyperplane that achieves the maximum margin (creates the widest possible street).

    By maximizing this cushion, SVM ensures the decision boundary stays as far away from both classes as possible, giving the model the maximum safety clearance when predicting unseen data points.

# Mathematical Example

1. **The Core Formulas**

    - Decision Function:
    $$f(x) = w_1 x_1 + w_2 x_2 + b$$

    - Prediction Rule:
    $$\hat{y} = \text{sign}(f(x)) = \begin{cases} +1 & \text{if } f(x) \ge 0 \\ -1 & \text{if } f(x) < 0 \end{cases}$$

    - Support Vector Boundary Constraint:
    $$y_i (w_1 x_1 + w_2 x_2 + b) \ge 1$$

    - Total Margin Width:
    $$\text{Margin} = \frac{2}{\Vert{}w\Vert{}} = \frac{2}{\sqrt{w_1^2 + w_2^2}}$$



2. **Dataset**

    Suppose we have 4 training data points with two features ($x_1, x_2$) and target labels $y \in \{-1, +1\}$:

    |Point|Feature x1|​Feature x2|​Class Label (y)|
    |---|---|---|---|
    |A|$2$|$3$|$+1$|
    |B|$3$|$3$|$+1$|
    |C|$1$|$1$|$-1$|
    |D|$2$|$1$|$-1$|


3. **Finding the Optimal Hyperplane ($w_1, w_2, b$)**

    Notice that as $x_1$ changes (from $1$ to $3$), the class label does not change at all. Only $x_2$ determines the class. Because $x_1$ has zero weight in separating these two clusters, its weight is $w_1 = 0$. The separating line must be purely horizontal.


    **Set Up the Support Vector Margin Equations**

    - Top Boundary (where $y = +1$ and $x_2 = 3$):
    $$w_1 x_1 + w_2 x_2 + b = +1$$
        
    - Bottom Boundary (where $y = -1$ and $x_2 = 1$):
    $$w_1 x_1 + w_2 x_2 + b = -1$$

    Substitute $w_1 = 0$ into both equations:
    1. Equation 1: $3w_2 + b = 1$
    2. Equation 2: $1w_2 + b = -1$

    Subtract Equation 2 from Equation 1 to eliminate $b$:
    $$(3w_2 + b) - (w_2 + b) = 1 - (-1)$$
    $$2w_2 = 2$$
    $$\mathbf{w_2 = 1}$$
    Now substitute $w_2 = 1$ back into Equation 2:
    $$1(1) + b = -1$$
    $$1 + b = -1$$
    $$\mathbf{b = -2}$$

    Setting $w_1 = 0$, $w_2 = 1$, and $b = -2$ gives the line equation $x_2 - 2 = 0$.


    **Testing the Constraint Formula on Training Data:**

    - Point A $(2, 3), y = +1$:
    $$y_A \cdot f(x_A) = (+1) \cdot (0(2) + 1(3) - 2) = (+1) \cdot (1) = \mathbf{1}$$
    - Point B $(3, 3), y = +1$:
    $$y_B \cdot f(x_B) = (+1) \cdot (0(3) + 1(3) - 2) = (+1) \cdot (1) = \mathbf{1}$$
    - Point C $(1, 1), y = -1$:
    $$y_C \cdot f(x_C) = (-1) \cdot (0(1) + 1(1) - 2) = (-1) \cdot (-1) = \mathbf{1}$$
    - Point D $(2, 1), y = -1$:
    $$y_D \cdot f(x_D) = (-1) \cdot (0(2) + 1(1) - 2) = (-1) \cdot (-1) = \mathbf{1}$$

    Because $y_i \cdot f(x_i) = 1$ for all points, every training point sits directly on the boundary margins, making them all Support Vectors.


4. **Calculating the Physical Margin Width**
    Using $w = (0, 1)$:
    $$\Vert{}w\Vert{} = \sqrt{0^2 + 1^2} = 1$$
    $$\text{Margin Width} = \frac{2}{\Vert{}w\Vert{}} = \frac{2}{1} = \mathbf{2 \text{ units}}$$

    The street width separating class boundaries is exactly 2 units wide ($x_2 = 3$ down to $x_2 = 1$).

5. **Predicting Unseen Test Data**
    Using learned weights $w_1 = 0, w_2 = 1, b = -2$:

    - New Query Point 1: $x_{\text{test}} = (2, 4)$
    $$f(x) = 0(2) + 1(4) - 2 = +2$$
    
    $$\hat{y} = \text{sign}(+2) = \mathbf{+1} \quad \text{(Classified as Positive)}$$
    
    - New Query Point 2: $x_{\text{test}} = (0, 0)$
    $$f(x) = 0(0) + 1(0) - 2 = -2$$
    $$\hat{y} = \text{sign}(-2) = \mathbf{-1} \quad \text{(Classified as Negative)}$$




# Terminologies

1. **Hard Margin SVM (Zero Tolerance)**
    Hard Margin SVM forces a strict mathematical separation. It assumes your data is perfectly clean and linearly separable.

2. **Soft Margin SVM (Realistic Tolerance)**
    Soft Margin SVM introduces Slack Variables (ξ). It accepts that real-world data is messy, noisy, and has overlapping regions.

    **Slack Variables ($\xi_i$)**
    For every data point $x_i$, a slack variable $\xi_i \ge 0$ measures its distance from its correct margin boundary.
    $$\text{Modified Constraint: } y_i(w^T x_i + b) \ge 1 - \xi_i$$

    - $\xi_i = 0$: The point is correctly classified and lies outside or directly on its correct margin edge.
    - $0 < \xi_i \le 1$: The point is correctly classified, but lies inside the margin street (a margin violation).
    - $\xi_i > 1$: The point falls on the wrong side of the decision boundary (a misclassification error).

    Soft-Margin SVM balances maximizing the margin width ($\frac{2}{\Vert{}w\Vert{}}$) with minimizing total classification slack ($\sum \xi_i$):
    $$\min_{w, b, \xi} \left( \frac{1}{2} \Vert{}w\Vert{}^2 + C \sum_{i=1}^{N} \xi_i \right) \quad \text{subject to} \quad y_i(w^T x_i + b) \ge 1 - \xi_i, \quad \xi_i \ge 0$$

    - The C Parameter: This is the slider that controls how "soft" the margin is.
    - A Large C means heavy penalties for mistakes, making the margin behave almost like a Hard Margin (narrow street, fewer errors allowed).
    - A Small C means light penalties, creating a very Soft Margin (wide street, more crossing allowed for a smoother line).


3. **What is a Kernel & Why Do We Use It?**

    - **The Problem:** Standard SVM can only draw straight lines or flat planes. If your data forms an XOR pattern or a circle inside a ring, a straight line will fail.
    
    - **The Kernel Solution:** A kernel is a similarity function. Instead of trying to find a line equation, a kernel takes two data points ($x_1$ and $x_2$) and outputs a single number measuring how "similar" or close they are to each other.

4. **What is the "Kernel Trick"?**
    
    - To separate non-linear data linearly, you theoretically have to project 2D points $(x_1, x_2)$ into a higher 3D or 4D space (e.g., adding a $z = x_1^2 + x_2^2$ axis) where a flat plane can slice between them.
    
    - Mapping thousands of data points into higher (or infinite) dimensions requires massive memory and computation.
    - The Trick: You never actually convert the data points to higher dimensions! The kernel formula $K(x_1, x_2)$ takes your original low-dimensional points and computes the exact same mathematical result as if you had mapped them to higher dimensions and taken their dot product. It is a computational shortcut.

5. **The Radial Basis Function (RBF) Kernel**
    
    The RBF (or Gaussian) kernel is the most widely used non-linear kernel:
    $$K(x_i, x_j) = \exp\left(-\gamma \Vert{}x_i - x_j\Vert{}^2\right)$$
    
    - Similarity Metric: Measures Euclidean distance $\Vert{}x_i - x_j\Vert{}^2$. If two points are identical, $K(x_i, x_j) = e^0 = 1$. As distance grows, $K \to 0$.
    
    - Infinite Dimensions: Through the Taylor series expansion of $\exp(z)$, the RBF kernel implicitly projects input vectors into an infinite-dimensional Hilbert space.
    
    **Role of the $\gamma$ (Gamma) Hyperparameter**
    
    - High $\gamma$: Creates narrow Gaussian peaks around support vectors. Only nearby points influence prediction, producing complex, tight decision boundaries (risk of overfitting).
    
    - Low $\gamma$: Creates broad, smooth Gaussian peaks. Points far away influence prediction, producing smoother decision boundaries (risk of underfitting).

# Python Implemenation

## Linear Support Vector Classification (SVC)

```python
import numpy as np


class LinearSVC:

  def __init__(self, C=1.0, lr=0.001, epochs=1000):
    self.C = C
    self.lr = lr
    self.epochs = epochs
    self.w = None
    self.b = None

  def fit(self, X, y):
    # Ensure targets are {-1, +1}
    y_ = np.where(y <= 0, -1, 1)
    n_samples, n_features = X.shape

    self.w = np.zeros(n_features)
    self.b = 0.0

    for _ in range(self.epochs):
      for idx, x_i in enumerate(X):
        # Condition: y_i * (w^T * x_i + b) >= 1
        condition = y_[idx] * (np.dot(x_i, self.w) + self.b) >= 1

        if condition:
          # Point is correctly outside the margin
          dw = self.w
          db = 0.0
        else:
          # Point violates the margin
          dw = self.w - self.C * y_[idx] * x_i
          db = -self.C * y_[idx]

        self.w -= self.lr * dw
        self.b -= self.lr * db

    return self

  def predict(self, X):
    linear_output = np.dot(X, self.w) + self.b
    return np.where(linear_output >= 0, 1, 0)



# TEST SCRIPT: CLASSIFICATION

# 2D Binary Classification Dataset
X_train_cls = np.array(
    [[1, 2], [2, 3], [2, 1], [3, 2], [7, 8], [8, 9], [7, 6], [9, 7]]
)
y_train_cls = np.array([0, 0, 0, 0, 1, 1, 1, 1])

# Train Model
svc = LinearSVC(C=1.0, lr=0.01, epochs=1000)
svc.fit(X_train_cls, y_train_cls)

# Test Queries
X_test_cls = np.array([[1.5, 1.5], [8.0, 7.5]])
preds_cls = svc.predict(X_test_cls)

print("=== LINEAR SVC RESULTS ===")
print(f"Learned Weights: w = {np.round(svc.w, 4)}, b = {np.round(svc.b, 4)}")
for i, sample in enumerate(X_test_cls):
  print(f"Sample {sample} -> Predicted Class: {preds_cls[i]}")
```

## Linear Support Vector Regression (SVR)


```python
import numpy as np


class LinearSVR:

  def __init__(self, C=1.0, epsilon=0.1, lr=0.001, epochs=1000):
    self.C = C
    self.epsilon = epsilon
    self.lr = lr
    self.epochs = epochs
    self.w = None
    self.b = None

  def fit(self, X, y):
    n_samples, n_features = X.shape

    self.w = np.zeros(n_features)
    self.b = 0.0

    for _ in range(self.epochs):
      for idx, x_i in enumerate(X):
        y_pred = np.dot(x_i, self.w) + self.b
        error = y[idx] - y_pred

        # Check if error exceeds epsilon tube boundary
        if error > self.epsilon:
          dw = self.w - self.C * x_i
          db = -self.C
        elif error < -self.epsilon:
          dw = self.w + self.C * x_i
          db = self.C
        else:
          # Error sits inside tube; no gradient penalty applied
          dw = self.w
          db = 0.0

        self.w -= self.lr * dw
        self.b -= self.lr * db

    return self

  def predict(self, X):
    return np.dot(X, self.w) + self.b


# TEST SCRIPT: REGRESSION

# Synthetic Linear Data: y = 2*x1 + 3*x2 + 5 + noise
np.random.seed(42)
X_train_reg = np.array([[1, 2], [2, 1], [3, 4], [4, 3], [5, 6], [6, 5]])
y_train_reg = np.array([13.1, 12.0, 23.2, 22.1, 33.0, 31.8])

# Train Model
svr = LinearSVR(C=1.0, epsilon=0.2, lr=0.001, epochs=2000)
svr.fit(X_train_reg, y_train_reg)

# Test Queries
X_test_reg = np.array([[2, 2], [5, 5]])
preds_reg = svr.predict(X_test_reg)

print("\n=== LINEAR SVR RESULTS ===")
print(f"Learned Weights: w = {np.round(svr.w, 4)}, b = {np.round(svr.b, 4)}")
for i, sample in enumerate(X_test_reg):
  print(f"Sample {sample} -> Predicted Value: {round(preds_reg[i], 2)}")
```

## Non-Linear Support Vector Classification (Kernel SVC)

```python
import numpy as np


class KernelSVC:

  def __init__(self, C=1.0, gamma=0.5, lr=0.01, epochs=1000):
    self.C = C
    self.gamma = gamma
    self.lr = lr
    self.epochs = epochs
    self.alpha = None
    self.b = 0.0
    self.X_train = None
    self.y_train = None

  def _rbf_kernel(self, X1, X2):
    # Pairwise squared Euclidean distance: ||X1 - X2||^2
    sq_dists = (
        np.sum(X1**2, axis=1)[:, None]
        + np.sum(X2**2, axis=1)[None, :]
        - 2 * np.dot(X1, X2.T)
    )
    return np.exp(-self.gamma * sq_dists)

  def fit(self, X, y):
    self.X_train = np.asarray(X, dtype=np.float64)
    # Convert labels from {0, 1} to {-1, +1}
    self.y_train = np.where(y <= 0, -1.0, 1.0)
    n_samples = self.X_train.shape[0]

    # Precompute RBF Kernel Matrix K (N x N)
    K = self._rbf_kernel(self.X_train, self.X_train)

    # Initialize dual variables alpha_i
    self.alpha = np.zeros(n_samples)

    # Dual Projected Gradient Ascent
    for _ in range(self.epochs):
      # Gradient w.r.t alpha: 1 - y * (K @ (alpha * y))
      grad = 1.0 - self.y_train * (K @ (self.alpha * self.y_train))
      self.alpha += self.lr * grad
      # Box constraint enforcement: 0 <= alpha_i <= C
      self.alpha = np.clip(self.alpha, 0, self.C)

    # Compute Bias b using Support Vectors (0 < alpha < C)
    sv_mask = (self.alpha > 1e-5) & (self.alpha < self.C - 1e-5)
    if not np.any(sv_mask):
      sv_mask = self.alpha > 1e-5  # Fallback if all SVs hit boundary

    if np.any(sv_mask):
      self.b = np.mean(
          self.y_train[sv_mask] - (K[sv_mask] @ (self.alpha * self.y_train))
      )

    return self

  def predict(self, X):
    X = np.asarray(X, dtype=np.float64)
    # Kernel between test points and training support vectors
    K_test = self._rbf_kernel(X, self.X_train)
    decision_scores = K_test @ (self.alpha * self.y_train) + self.b
    return np.where(decision_scores >= 0, 1, 0)


# TEST SCRIPT: NON-LINEAR XOR DATASET

# Non-Linearly Separable XOR Pattern
X_train_xor = np.array([[-1, -1], [-1, 1], [1, -1], [1, 1]])
y_train_xor = np.array([0, 1, 1, 0])  # Opposite corners have same label

# Train Model
ksvc = KernelSVC(C=10.0, gamma=1.0, lr=0.01, epochs=2000)
ksvc.fit(X_train_xor, y_train_xor)

# Test Queries
X_test_xor = np.array([[-0.8, -0.9], [0.9, -0.7], [-0.7, 0.8], [0.8, 0.9]])
preds_xor = ksvc.predict(X_test_xor)

print("=== NON-LINEAR KERNEL SVC (XOR PROBLEM) RESULTS ===")
for i, sample in enumerate(X_test_xor):
  print(f"Sample {sample} -> Predicted Class: {preds_xor[i]}")
```

## Non-Linear Support Vector Regression (Kernel SVR)

```python
import numpy as np


class KernelSVR:

  def __init__(self, C=1.0, epsilon=0.1, gamma=0.5, lr=0.005, epochs=1500):
    self.C = C
    self.epsilon = epsilon
    self.gamma = gamma
    self.lr = lr
    self.epochs = epochs
    self.beta = None  # Dual weight vector (alpha* - alpha)
    self.b = 0.0
    self.X_train = None

  def _rbf_kernel(self, X1, X2):
    sq_dists = (
        np.sum(X1**2, axis=1)[:, None]
        + np.sum(X2**2, axis=1)[None, :]
        - 2 * np.dot(X1, X2.T)
    )
    return np.exp(-self.gamma * sq_dists)

  def fit(self, X, y):
    self.X_train = np.asarray(X, dtype=np.float64)
    y = np.asarray(y, dtype=np.float64)
    n_samples = self.X_train.shape[0]

    K = self._rbf_kernel(self.X_train, self.X_train)
    self.beta = np.zeros(n_samples)

    # Gradient Descent on Dual SVR Formulation
    for _ in range(self.epochs):
      y_pred = (K @ self.beta) + self.b
      errors = y_pred - y

      # Subgradient for epsilon-insensitive tube
      tube_grad = np.where(
          errors > self.epsilon, 1.0, np.where(errors < -self.epsilon, -1.0, 0.0)
      )

      # Update dual weights and bias
      dbeta = (K @ self.beta) + (self.C * (K @ tube_grad))
      db = self.C * np.sum(tube_grad)

      self.beta -= self.lr * dbeta
      self.b -= self.lr * db

      # Constrain dual weights to [-C, +C]
      self.beta = np.clip(self.beta, -self.C, self.C)

    return self

  def predict(self, X):
    X = np.asarray(X, dtype=np.float64)
    K_test = self._rbf_kernel(X, self.X_train)
    return (K_test @ self.beta) + self.b


# =====================================================================
# TEST SCRIPT: NON-LINEAR QUADRATIC CURVE (y = x^2)
# =====================================================================

# 1D Non-Linear Data: Quadratic Curve y = x^2
X_train_curve = np.array([[-2.0], [-1.5], [-1.0], [0.0], [1.0], [1.5], [2.0]])
y_train_curve = np.array([4.0, 2.25, 1.0, 0.0, 1.0, 2.25, 4.0])

# Train Model
ksvr = KernelSVR(C=5.0, epsilon=0.1, gamma=0.5, lr=0.005, epochs=2000)
ksvr.fit(X_train_curve, y_train_curve)

# Test Queries (Predict points on the curve)
X_test_curve = np.array([[-1.8], [0.5], [1.8]])
preds_curve = ksvr.predict(X_test_curve)

print("\n=== NON-LINEAR KERNEL SVR (y = x^2) RESULTS ===")
for i, sample in enumerate(X_test_curve):
  actual_expected = sample[0] ** 2
  print(
      f"Query x = {sample[0]} -> Predicted y: {round(preds_curve[i], 2)} "
      f"(Actual x^2: {actual_expected})"
  )
```

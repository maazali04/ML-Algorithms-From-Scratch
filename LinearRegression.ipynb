# Simple Linear Regression Model


Simple Linear Regression is a statistical method used to model and analyze the relationship between two variables: one independent variable (the cause) and one dependent variable (the effect)

As we know the line equation as:

$$y = mx +b$$

where
- x (Input/Feature): The value you know.
- m : (slope / Weight) representing how steep the line is. It tells you "If x increases by 1, how much does y change?"
- b (intercept / bias) : Controels where the line crosses the veritical y-axis.
- y (Output / Target): The value you want to predict

For a single data point ($X_i ,y_i$), our line predicts a vlue called $ \hat y_i $:
$$ \hat y_i = mX_i + b$$

**Loss**

To evaluate our whole dataset of $N$ points, we need a single number that tells us how bad our line is. For that purpose we use Mean Squared Error (MSE).
$$\text{MSE} = \frac{1}{N} \sum_{i=1}^{N} (\hat{y}_i - y_i)^2$$

**Why not raw sum of errors? $\sum (\hat{y}_i - y_i)$**

If one point has an error of $+5$ and another has an error of $-5$, adding them together gives $0$. The errors cancel each other out, making a terrible line look perfect! We need all errors to be positive numbers.

**Why not Absolute Error (MAE)? $\frac{1}{N} \sum \vert{}\hat{y}_i - y_i\vert{}$**

Absolute value turns negative numbers positive, which solves the cancellation problem. However, mathematically, the absolute value function creates a sharp "V" shape with a pointy corner at $0$. You cannot take a smooth derivative (calculus rate of change) at a sharp point. Smooth derivatives are required for algorithms like Gradient Descent to run efficiently.

**Why the "Mean" ($\frac{1}{N}$)?**

If you only take the Sum of Squared Errors, a dataset with $1,000$ points will naturally have a much larger error total than a dataset with $10$ points, even if the line fits both equally well. Dividing by $N$ (the average) ensures your error metric stays comparable regardless of dataset size.

**The Big Picture: Closed-Form vs. Gradient Descent**

In linear regression, there are two ways to get $m$ and $b$:

**Analytical Solution (OLS):** There are no "updates" or "learning rates" here because you plug in the numbers once and get the final answer.

**Gradient Descent (Iterative Optimization):** We start with total random guesses for $m$ and $b$, and we modify their values step-by-step over time to minimize the error.

## OLS

**Finding the Best $m$ and $b$**

We want to find the exact values of $m$ and $b$ that make MSE as small as possible.

Our Loss Function is:
$$\text{MSE} = \frac{1}{N} \sum_{i=1}^{N} (\hat{y}_i - y_i)^2$$
put the values of $\hat y_i (\hat y_i = m X_i + b)$
$$J(m, b) = \frac{1}{N} \sum_{i=1}^{N} (m X_i + b - y_i)^2$$

Part A: Finding Optimal $b$ (Bias)

Take the partial derivative of $J(m, b)$ with respect to $b$:$$\frac{\partial J}{\partial b} = \frac{2}{N} \sum_{i=1}^{N} (m X_i + b - y_i)$$Set the derivative to $0$ to find the minimum:$$\frac{2}{N} \sum_{i=1}^{N} (m X_i + b - y_i) = 0$$Divide both sides by $2$:$$\frac{1}{N} \sum_{i=1}^{N} (m X_i + b - y_i) = 0$$Split the sum across the terms:$$m \left(\frac{1}{N} \sum X_i\right) + \left(\frac{1}{N} \sum b\right) - \left(\frac{1}{N} \sum y_i\right) = 0$$Notice that $\frac{1}{N} \sum X_i$ is just the average of $X$ ($\bar{X}$), and $\frac{1}{N} \sum y_i$ is the average of $y$ ($\bar{y}$). Also, adding $b$ to itself $N$ times and dividing by $N$ just equals $b$.Substitute these averages into the equation:$$m \bar{X} + b - \bar{y} = 0$$Rearrange to solve for $b$:$$b = \bar{y} - m \bar{X}$$What this means intuitively: The optimal y-intercept $b$ guarantees that your regression line always passes through the exact center point $(\bar{X}, \bar{y})$ of your dataset!


Part B: Finding Optimal $m$ (Slope)

Take the partial derivative of $J(m, b)$ with respect to $m$ using the chain rule:$$\frac{\partial J}{\partial m} = \frac{2}{N} \sum_{i=1}^{N} X_i (m X_i + b - y_i)$$Set it equal to $0$ and divide by $2$:$$\sum_{i=1}^{N} X_i (m X_i + b - y_i) = 0$$Substitute our formula for $b = \bar{y} - m \bar{X}$ into this equation:$$\sum_{i=1}^{N} X_i \Big( m X_i + (\bar{y} - m \bar{X}) - y_i \Big) = 0$$Group the $m$ terms together:$$\sum_{i=1}^{N} X_i \Big( m(X_i - \bar{X}) - (y_i - \bar{y}) \Big) = 0$$Expand the brackets:$$m \sum_{i=1}^{N} X_i(X_i - \bar{X}) - \sum_{i=1}^{N} X_i(y_i - \bar{y}) = 0 $$  


The Essential Hint: Why $\sum (y_i - \bar{y}) = 0$

Take any list of numbers, calculate their average ($\bar{y}$), and sum up how far each number is from that average:$$\sum_{i=1}^{N} (y_i - \bar{y}) = \sum_{i=1}^{N} y_i - \sum_{i=1}^{N} \bar{y} = \sum_{i=1}^{N} y_i - N\bar{y}$$By definition, $\bar{y} = \frac{1}{N}\sum y_i$, so $N\bar{y} = \sum y_i$:$$\sum_{i=1}^{N} y_i - \sum_{i=1}^{N} y_i = 0$$Rule: The sum of deviations from the mean is always zero. Multiply it by any constant like $\bar{X}$, and $\bar{X} \sum (y_i - \bar{y})$ is still strictly 0.


Transforming the Numerator:

Let's look at $\sum (X_i - \bar{X})(y_i - \bar{y})$. Expand it:

$$\sum_{i=1}^{N} (X_i - \bar{X})(y_i - \bar{y}) = \sum_{i=1}^{N} \Big( X_i(y_i - \bar{y}) - \bar{X}(y_i - \bar{y}) \Big)$$

$$\sum_{i=1}^{N} (X_i - \bar{X})(y_i - \bar{y}) = \sum_{i=1}^{N} X_i(y_i - \bar{y}) - \bar{X} \underbrace{\sum_{i=1}^{N} (y_i - \bar{y})}_{= 0}$$

$$\sum_{i=1}^{N} (X_i - \bar{X})(y_i - \bar{y}) = \sum_{i=1}^{N} X_i(y_i - \bar{y}) - 0$$
$$\mathbf{\sum_{i=1}^{N} X_i(y_i - \bar{y}) = \sum_{i=1}^{N} (X_i - \bar{X})(y_i - \bar{y})}$$


Transforming the Denominator:

Apply the exact same expansion to $\sum (X_i - \bar{X})^2$:

$$\sum_{i=1}^{N} (X_i - \bar{X})^2 = \sum_{i=1}^{N} (X_i - \bar{X})(X_i - \bar{X}) = \sum_{i=1}^{N} X_i(X_i - \bar{X}) - \bar{X} \underbrace{\sum_{i=1}^{N} (X_i - \bar{X})}_{= 0}$$

$$\mathbf{\sum_{i=1}^{N} X_i(X_i - \bar{X}) = \sum_{i=1}^{N} (X_i - \bar{X})^2}$$


**The Final Formulas**

Substituting these simplified terms into Equation 1 gives the standard Ordinary Least Squares (OLS) formula: $$\mathbf{m = \frac{\sum_{i=1}^{N} (X_i - \bar{X})(y_i - \bar{y})}{\sum_{i=1}^{N} (X_i - \bar{X})^2}}$$
$$\mathbf{b = \bar{y} - m \bar{X}}$$

```python
import numpy as np

class OLS_SLR_Model():

    def __init__(self):
        self.m = None
        self.b = None

    def fit(self, X, y):
        self.X = np.array(X)
        self.Y = np.array(y)
        
        self.x_mean = np.mean(self.X)
        self.y_mean = np.mean(self.Y)

        numerator = np.sum((self.X - self.x_mean) * (self.Y - self.y_mean))
        denominator = np.sum((self.X - self.x_mean)**2)

        self.m = numerator / denominator
        self.b = self.y_mean - (self.m * self.x_mean)
        
        return self

    def predict(self, X):
        X = np.array(X)
        return self.m * X + self.b
```

```python
model = OLS_SLR_Model()
model.fit([1, 2, 3], [2, 4, 6])

print(f"Slope (m): {model.m}")       
print(f"Intercept (b): {model.b}")   
print(f"Prediction for X=4: {model.predict([4])}") 
```

# Multiple Linear Regression Model

In Multiple Linear Regression, instead of fitting a 2D line using one input feature ($X$), we fit an $n$-dimensional hyperplane using $n$ different features ($X_1, X_2, \dots, X_n$).

**The Model Equation**

For a single data point with $p$ features, our hypothesis equation looks like this:$$\hat{y} = w_0 + w_1 X_1 + w_2 X_2 + \dots + w_p X_p$$
Where:

$w_0$ is the intercept (bias term, previously called $b$).

$w_1, w_2, \dots, w_p$ are the weights/coefficients for each feature (previously $m$).

Let's look at the individual hypothesis equations for *N* different data points, each with *P* features:
$$\hat{y}_{1}=w_{0}(1)+w_{1}x_{11}+w_{2}x_{12}+\dots +w_{p}x_{1p}$$
$$\hat{y}_{2}=w_{0}(1)+w_{1}x_{21}+w_{2}x_{22}+\dots +w_{p}x_{2p}$$
$$\vdots$$
$$\hat{y}_{N}=w_{0}(1)+w_{1}x_{31}+w_{2}x_{32}+\dots +w_{p}x_{3p}$$

Applying the row-by-column rule gives us:

$$ {\mathbf{Y}}=\mathbf{XW}=\left[\begin{matrix}1&x_{11}&x_{12}&\dots &x_{1p}\\ 1&x_{21}&x_{22}&\dots &x_{2p}\\ \vdots &\vdots &\vdots &\ddots &\vdots \\ 1&x_{N1}&x_{N2}&\dots &x_{Np}\end{matrix}\right]\left[\begin{matrix}w_{0}\\ w_{1}\\ \vdots \\ w_{p}\end{matrix}\right]$$

$$\mathbf{Y} = \begin{bmatrix} y_1 \\ y_2 \\ \vdots \\ y_N \end{bmatrix}, \quad \mathbf{W} = \begin{bmatrix} w_0 \\ w_1 \\ \vdots \\ w_p \end{bmatrix}, \quad \mathbf{X} = \begin{bmatrix} 1 & x_{11} & x_{12} & \dots & x_{1p} \\ 1 & x_{21} & x_{22} & \dots & x_{2p} \\ \vdots & \vdots & \vdots & \ddots & \vdots \\ 1 & x_{N1} & x_{N2} & \dots & x_{Np} \end{bmatrix}$$

**Loss Function**

Let's suppose:
    $$ e = y_i - \hat y_i$$
If we pack the errors for all *N* data points into a single column vector, we get the **Error Vector $ \mathbf{E}$ :**

$$\mathbf{E}=\mathbf{Y}-\^{\mathbf{Y}}=\left[\begin{matrix}y_{1}-\^{y}_{1}\\ y_{2}-\^{y}_{2}\\ \vdots \\ y_{N}-\^{y}_{N}\end{matrix}\right]=\left[\begin{matrix}e_{1}\\ e_{2}\\ \vdots \\ e_{N}\end{matrix}\right]$$

Your original loss function $J(\mathbf{W})$ is simply the sum of these squared errors:

$$J(\mathbf{W})=\sum _{i=1}^{N}(y_{i}-\^{y}_{i})^{2}=e_{1}^{2}+e_{2}^{2}+\dots +e_{N}^{2} = \sum _{i=1}^N e_{i}^2$$

Now let's look at what happens when you take the transpose of this column vector $\mathbf{E}$ (which turns it into a horizontal row) and multiply it by the original vertical vector $\mathbf{E}$.

$$\mathbf{E}^{T}\mathbf{E}=\left[\begin{matrix}e_{1}&e_{2}&\dots &e_{N}\end{matrix}\right]\left[\begin{matrix}e_{1}\\ e_{2}\\ \vdots \\ e_{N}\end{matrix}\right]$$

$$ \mathbf{E}^{T}\mathbf{E}=(e_{1}\cdot e_{1})+(e_{2}\cdot e_{2})+\dots +(e_{N}\cdot e_{N})$$
$$\mathbf{E}^{T}\mathbf{E}=e_{1}^{2}+e_{2}^{2}+\dots +e_{N}^{2}$$
$$\mathbf{E}^{T}\mathbf{E}=\sum _{i=1}^N e_{i}^2 $$


Since we defined $\mathbf{E}$ as $(\mathbf{Y} - \mathbf{\hat{Y}})$, we simply replace $\mathbf{E}$ with the full vector expression:
$$J(\mathbf{W}) = \sum _{i=1}^{N}(y_{i}-\^{y}_{i})^{2}=\mathbf{E}^{T}\mathbf{E}=(\mathbf{Y}-\^{\mathbf{Y}})^{T}(\mathbf{Y}-\^{\mathbf{Y}})$$

Substitute $\mathbf{\hat{Y}} = \mathbf{XW}$:
$$J(\mathbf{W}) = (\mathbf{Y} - \mathbf{XW})^T (\mathbf{Y} - \mathbf{XW})$$


Using matrix transposition rules $(A - B)^T = A^T - B^T$ and $(AB)^T = B^T A^T$:

$$J(\mathbf{W}) = (\mathbf{Y}^T - \mathbf{W}^T \mathbf{X}^T) (\mathbf{Y} - \mathbf{XW})$$
$$J(\mathbf{W}) = \mathbf{Y}^T \mathbf{Y} - \mathbf{Y}^T \mathbf{XW} - \mathbf{W}^T \mathbf{X}^T \mathbf{Y} + \mathbf{W}^T \mathbf{X}^T \mathbf{XW}$$

Notice that $\mathbf{Y}^T \mathbf{XW}$ are identical:
$$\mathbf{Y}^T \mathbf{XW} = \mathbf{W}^T \mathbf{X}^T \mathbf{Y}$$

So we combine them into $-2 \mathbf{W}^T \mathbf{X}^T \mathbf{Y}$:
$$\mathbf{J(\mathbf{W}) = \mathbf{Y}^T \mathbf{Y} - 2 \mathbf{W}^T \mathbf{X}^T \mathbf{Y} + \mathbf{W}^T \mathbf{X}^T \mathbf{XW}}$$

To find the weight vector $\mathbf{W}$ that minimizes error, we take the gradient (matrix derivative) of $J(\mathbf{W})$ with respect to $\mathbf{W}$ and set it to 0.

Now differentiate each term in $J(\mathbf{W})$ with respect to $\mathbf{W}$:
1. $\frac{\partial}{\partial \mathbf{W}} (\mathbf{Y}^T \mathbf{Y}) = \mathbf{0}$  (No $\mathbf{W}$ in this term)
2. $\frac{\partial}{\partial \mathbf{W}} (-2 \mathbf{W}^T \mathbf{X}^T \mathbf{Y}) = -2 \mathbf{X}^T \mathbf{Y}$
3. $\frac{\partial}{\partial \mathbf{W}} (\mathbf{W}^T \mathbf{X}^T \mathbf{XW}) = 2 \mathbf{X}^T \mathbf{XW}$

Combine them to get the gradient:
$$\frac{\partial J}{\partial \mathbf{W}} = -2 \mathbf{X}^T \mathbf{Y} + 2 \mathbf{X}^T \mathbf{XW}$$

Set the derivative to 0:
$$-2 \mathbf{X}^T \mathbf{Y} + 2 \mathbf{X}^T \mathbf{XW} = \mathbf{0}$$
$$ \mathbf{X}^T \mathbf{Y} +  \mathbf{X}^T \mathbf{XW} = \mathbf{0}$$
$$    \mathbf{X}^T \mathbf{XW} = \mathbf{X}^T \mathbf{Y}$$

$$\underbrace{(\mathbf{X}^T \mathbf{X})^{-1} (\mathbf{X}^T \mathbf{X})}_{\text{Identity Matrix } \mathbf{I}} \mathbf{W} = (\mathbf{X}^T \mathbf{X})^{-1} \mathbf{X}^T \mathbf{Y}$$

$$\mathbf{W = (\mathbf{X}^T \mathbf{X})^{-1} \mathbf{X}^T \mathbf{Y}}$$

```python
import numpy as np

class OLS_MLR_Model():
    def __init__(self):
        self.weights = None
    
    def fit(self, X, y):
        X = np.array(X)
        y = np.array(y)

        # Add intercept column (column of 1s) to X
        ones = np.ones((X.shape[0], 1))
        X_b = np.hstack([ones, X])

        # Transpose X_b
        X_transpose = X_b.T

        # W = (X^T @ X)^-1 @ X^T @ y
        self.weights = np.linalg.inv(X_transpose @ X_b) @ X_transpose @ y
        
        return self

    def predict(self, X):
        X = np.array(X)
        
        # Add intercept column (column of 1s) to test data as well
        ones = np.ones((X.shape[0], 1))
        X_b = np.hstack([ones, X])

        #  Y_pred = X_b @ W
        return X_b @ self.weights
```

```python
# Sample data: 2 features
X = np.array([
    [1, 2],
    [2, 1],
    [3, 4],
    [4, 3]
])

# Target: y = 5 + 3*(x1) + 2*(x2)
y = np.array([12, 13, 22, 23])

model = OLS_MLR_Model()
model.fit(X, y)

print("Learned Weights [w0, w1, w2]:", np.round(model.weights, 4))
pred = model.predict([[5, 5]])
print("Prediction for [5, 5]:", pred[0])
```

**Why OLS Isn't Enough:** 

The Matrix Inverse ProblemThe OLS Normal Equation requires calculating $(\mathbf{X}^T \mathbf{X})^{-1}$:
1. Computational Cost: Matrix inversion takes $O(p^3)$ time, where $p$ is the number of features. If you have $100,000$ features (e.g., text embeddings or genetics data), calculating $(\mathbf{X}^T \mathbf{X})^{-1}$ will freeze your computer or run out of memory.
2. Non-Invertible Matrices: If two features are perfectly correlated (multicollinearity) or if you have more features than samples ($p > N$), $(\mathbf{X}^T \mathbf{X})$ cannot be inverted at all!

This is where Gradient Descent comes in. It finds the exact same optimal weights without ever computing a matrix inverse.

## The Vectorized Gradient Descent Derivation

Instead of treating $m$ and $b$ separately, we combine them into a single weight vector $\mathbf{W}$:
$$\mathbf{W} = \begin{bmatrix} w_0 \\ w_1 \\ \vdots \\ w_p \end{bmatrix}, \quad \mathbf{X} = \begin{bmatrix} 1 & x_{11} & \dots & x_{1p} \\ 1 & x_{21} & \dots & x_{2p} \\ \vdots & \vdots & \ddots & \vdots \\ 1 & x_{N1} & \dots & x_{Np} \end{bmatrix}$$

**Vectorized Predictions ($\mathbf{\hat{Y}}$)**

For all $N$ data points at once:
$$\mathbf{\hat{Y}} = \mathbf{X} \mathbf{W}$$

**Loss Function (MSE)**
$$J(\mathbf{W}) = \frac{1}{N} (\mathbf{X}\mathbf{W} - \mathbf{Y})^T (\mathbf{X}\mathbf{W} - \mathbf{Y})$$

**Gradient Descent**
$$\frac{\partial J}{\partial \mathbf{W}} = \frac{1}{N} (-2 \mathbf{X}^T \mathbf{Y} + 2 \mathbf{X}^T \mathbf{XW})$$

Factor out the common multiplier $2\mathbf{X}^T$:

$$\nabla _{\mathbf{W}}J(\mathbf{W})=\frac{2}{N}\mathbf{X}^{T}(\mathbf{XW}-\mathbf{Y})$$
$$\nabla _{\mathbf{W}}J(\mathbf{W})=\frac{2}{N} \mathbf{X}^T (\mathbf{\hat{Y}} - \mathbf{Y})$$


**Vectorized Weight Update Rule**
At each iteration (epoch), we update the entire weight vector in one single line of matrix subtraction:
$$\mathbf{W} = \mathbf{W} - \alpha \cdot \frac{2}{N} \mathbf{X}^T (\mathbf{\hat{Y}} - \mathbf{Y})$$
(where $\alpha$ is the learning rate).

```python
import numpy as np

class GD_MLR_Model():
    def __init__(self, epochs=1000, lr=0.01):
        self.weights = None
        self.epochs = epochs
        self.lr = lr
    
    def fit(self, X, y):
        X = np.atleast_2d(X)
        y = np.array(y)
        n = X.shape[0]

        ones = np.ones((n, 1))
        X_b = np.hstack([ones, X])

        # Initialize weights array to zeros matching number of columns in X_b
        self.weights = np.zeros(X_b.shape[1])

        X_transpose = X_b.T 

        for epoch in range(self.epochs):
            y_pred = X_b @ self.weights
            gradient = (2 / n) * (X_transpose @ (y_pred - y))
            self.weights = self.weights - (self.lr * gradient)

        return self

    def predict(self, X):
        X = np.atleast_2d(X)
        ones = np.ones((X.shape[0], 1))
        X_b = np.hstack([ones, X])
        return X_b @ self.weights
```

```python
import numpy as np

# Raw unscaled data
X = np.array([
    [1, 2],
    [2, 1],
    [3, 4],
    [4, 3],
    [5, 5]
])

# Target: y = 4 + 2*(x1) + 5*(x2)
y = np.array([16, 13, 30, 27, 39])

# Train directly without feature engineering / scaling
model = GD_MLR_Model(epochs=10000, lr=0.01)
model.fit(X, y)

print("Learned Weights [w0, w1, w2]:", np.round(model.weights, 4))
```

# Polynomial Regression

Polynomial Regression is a form of linear regression where the relationship between the independent variable \(X\) and the dependent variable \(y\) is modeled as an \(n\)-th degree polynomial

## 2-Degree Polynomial Regression (single variable)
Suppose your target $y$ follows a quadratic curve (like a parabola):$$y = w_0 + w_1 x + w_2 x^2$$

At first glance, this equation looks non-linear because $x$ is squared ($x^2$).However, in machine learning, "linear" refers to the parameters/weights ($W$), NOT the input features ($X$).

Let $z_1 = x$ and  $z_2 = x^2$
$$y = w_0 + w_1 z_1 + w_2 z_2$$

This is identical to Multiple Linear Regression with two features ($z_1$ and $z_2$)!

**The Equation**

Let’s derive the math for a single variable ($x$), degree-2 polynomial (a parabola).

Instead of fitting a straight line ($y = m x + b$), we are fitting a quadratic curve. We have one input variable $x$, but we need to learn three weights ($w_0, w_1, w_2$):
$$y = w_0 + w_1 x + w_2 x^2$$
- $w_0$ is the intercept.
- $w_1$ is the linear weight.
- $w_2$ is the quadratic weight (controls the curve).


**The Loss Function**
$$J(w_0, w_1, w_2) = \sum_{i=1}^{N} \left( y_i - (w_0 + w_1 x_i + w_2 x_i^2) \right)^2$$

**Taking the Derivatives**

1. With respect to $w_0$:
$$\frac{\partial J}{\partial w_0} = -2 \sum \left( y_i - (w_0 + w_1 x_i + w_2 x_i^2) \right) \times (1) = 0$$

2. With respect to $w_1$:
$$\frac{\partial J}{\partial w_1} = -2 \sum \left( y_i - (w_0 + w_1 x_i + w_2 x_i^2) \right) \times (x_i) = 0$$

3. With respect to $w_2$:
$$\frac{\partial J}{\partial w_2} = -2 \sum \left( y_i - (w_0 + w_1 x_i + w_2 x_i^2) \right) \times (x_i^2) = 0$$

**The System of Equations**

$$\sum y = N w_0 + w_1 \sum x + w_2 \sum x^2$$
$$\sum xy = w_0 \sum x + w_1 \sum x^2 + w_2 \sum x^3$$
$$\sum x^2y = w_0 \sum x^2 + w_1 \sum x^3 + w_2 \sum x^4$$

Let's Convert them into Matrix
$$\sum x^2y = w_0 \sum x^2 + w_1 \sum x^3 + w_2 \sum x^4$$

Let's Convert them into Matrix

$$ \left[\begin{matrix}N w_0 + w_1 \sum x + w_2 \sum x^2 \\ w_0 \sum x + w_1 \sum x^2 + w_2 \sum x^3 \\ w_0 \sum x^2 + w_1 \sum x^3 + w_2 \sum x^4\end{matrix}\right]=\left[\begin{matrix}\sum y\\ \sum xy\\ \sum x^{2}y\end{matrix}\right]$$


$$\left[\begin{matrix}N&\sum x&\sum x^{2}\\ \sum x&\sum x^{2}&\sum x^{3}\\ \sum x^{2}&\sum x^{3}&\sum x^{4}\end{matrix}\right]\left[\begin{matrix}w_{0}\\ w_{1}\\ w_{2}\end{matrix}\right]=\left[\begin{matrix}\sum y\\ \sum xy\\ \sum x^{2}y\end{matrix}\right]$$

$$A = \begin{bmatrix} N & \sum x & \sum x^2 \\ \sum x & \sum x^2 & \sum x^3 \\ \sum x^2 & \sum x^3 & \sum x^4 \end{bmatrix}, \quad W = \begin{bmatrix} w_0 \\ w_1 \\ w_2 \end{bmatrix}, \quad B = \begin{bmatrix} \sum y \\ \sum xy \\ \sum x^2y \end{bmatrix}$$

$$A W = B$$
$$W = A^{-1} B$$

```python
import numpy as np 

class Degree2_PolynomialRegression():
    def __init__(self):
        self.weights = None

    def fit(self, X, y):
        x = np.array(X)
        y = np.array(y)
        N = x.shape[0]

        sum_x  = np.sum(x)
        sum_x2 = np.sum(x**2)
        sum_x3 = np.sum(x**3)
        sum_x4 = np.sum(x**4)

        sum_y   = np.sum(y)
        sum_xy  = np.sum(x * y)
        sum_x2y = np.sum((x**2) * y)

        A = np.array([
            [N,      sum_x,  sum_x2],
            [sum_x,  sum_x2, sum_x3],
            [sum_x2, sum_x3, sum_x4]
        ])
        
        B = np.array([[sum_y], [sum_xy], [sum_x2y]])
        
        # W = A^-1 @ B
        self.weights = np.linalg.inv(A) @ B

        return self
    
    def predict(self, X):
        x = np.array(X)
        w0, w1, w2 = self.weights.ravel()
        return w0 + (w1 * x) + (w2 * (x**2))
```

```python
# y = 3 + 2x + x^2
x_data = [1, 2, 3]
y_data = [6, 11, 18]

model = Degree2_PolynomialRegression()
model.fit(x_data, y_data)

print("Learned Weights:\n", model.weights.ravel())

# y = 3 + 2*4 + 4^2 = 27
print("\nPrediction for X=4:", model.predict([4])[0])
```

## N-Degree polynomial Regression (single variable)

$$y_i = w_0 + w_1 x_i + w_2 x_i^2 + w_3 x_i^3 + \dots + w_d x_i^d$$

Expanding for all N Data Points

$$\begin{aligned} y_1 &= w_0 + w_1 x_1 + w_2 x_1^2 + w_3 x_1^3 + \dots + w_d x_1^d \\ y_2 &= w_0 + w_1 x_2 + w_2 x_2^2 + w_3 x_2^3 + \dots + w_d x_2^d \\ \vdots \\ y_N &= w_0 + w_1 x_N + w_2 x_N^2 + w_3 x_N^3 + \dots + w_d x_N^d \end{aligned}$$

Converting to Matrix Form

$$\begin{bmatrix}  w_0 + w_1 x_1 + w_2 x_1^2 + w_3 x_1^3 + \dots + w_d x_1^d \\ w_0 + w_1 x_2 + w_2 x_2^2 + w_3 x_2^3 + \dots + w_d x_2^d \\ \vdots \\  w_0 + w_1 x_N + w_2 x_N^2 + w_3 x_N^3 + \dots + w_d x_N^d  \end{bmatrix}$$



$$\begin{bmatrix} \hat{y}_1 \\ \hat{y}_2 \\ \vdots \\ \hat{y}_N \end{bmatrix} = \begin{bmatrix}  1 & x_1 & x_1^2 & x_1^3 & \dots & x_1^d \\  1 & x_2 & x_2^2 & x_2^3 & \dots & x_2^d \\  \vdots & \vdots & \vdots & \vdots & \ddots & \vdots \\  1 & x_N & x_N^2 & x_N^3 & \dots & x_N^d  \end{bmatrix} \begin{bmatrix} w_0 \\ w_1 \\ w_2 \\ w_3 \\ \vdots \\ w_d \end{bmatrix}$$

$$\mathbf{\hat{Y}} = \mathbf{X}_{\text{poly}} \mathbf{W}$$

Loss Function (Sum of Squared Errors)
$$J(\mathbf{W}) = \sum_{i=1}^{N} \left( y_i - \hat{y}_i \right)^2 = (\mathbf{Y} - \mathbf{X}_{\text{poly}}\mathbf{W})^T (\mathbf{Y} - \mathbf{X}_{\text{poly}}\mathbf{W})$$

Expanding the vector transpose multiplication:$$J(\mathbf{W}) = \mathbf{Y}^T \mathbf{Y} - 2 \mathbf{W}^T \mathbf{X}_{\text{poly}}^T \mathbf{Y} + \mathbf{W}^T \mathbf{X}_{\text{poly}}^T \mathbf{X}_{\text{poly}} \mathbf{W}$$

Matrix Derivative & Setting to Zero
$$\frac{\partial J}{\partial \mathbf{W}} = -2 \mathbf{X}_{\text{poly}}^T \mathbf{Y} + 2 \mathbf{X}_{\text{poly}}^T \mathbf{X}_{\text{poly}} \mathbf{W} = \mathbf{0}$$Dividing by 2 and moving terms:$$\mathbf{X}_{\text{poly}}^T \mathbf{X}_{\text{poly}} \mathbf{W} = \mathbf{X}_{\text{poly}}^T \mathbf{Y}$$

$$\mathbf{W} = (\mathbf{X}_{\text{poly}}^T \mathbf{X}_{\text{poly}})^{-1} \mathbf{X}_{\text{poly}}^T \mathbf{Y}$$


```python
import numpy as np 

class ND_PolynomialRegression():
    def __init__(self, degree=2):
        self.degree = degree
        self.weights = None

    def _transform_features(self, X):
        x = np.array(X).ravel()
        # Generates columns: [x^0, x^1, x^2, ..., x^degree]
        return np.column_stack([x**d for d in range(self.degree + 1)])

    def fit(self, X, y):
        y = np.array(y)
        X_poly = self._transform_features(X)
        X_poly_T = X_poly.T 
        self.weights = np.linalg.inv(X_poly_T @ X_poly) @ X_poly_T @ y
        return self
    
    def predict(self, X):
        X_poly = self._transform_features(X)
        return X_poly @ self.weights
```
